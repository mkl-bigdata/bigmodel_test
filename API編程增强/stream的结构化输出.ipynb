```python
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()
```

```python
client = OpenAI(
    api_key=os.environ['api_key'],
    base_url=os.environ['base_url'],
)
```

```python
from pydantic import BaseModel

class entities(BaseModel):
    color: str
    entity: str

with client.responses.stream(
    model="deepseek-v4-pro",
    input= [
        {"role": "system", "content": "Extract entities from the input text"},
        {
            "role": "user",
            "content": "The quick brown fox jumps over the lazy dog with piercing blue eyes",
        },
    ],
    text_format=entities
) as stream:
    for record in stream:
        if record.type == "response.output_text.delta":
            print(record.delta, end="", flush=True)
        elif record.type == "response.completed":
            print("\n\nResponse completed")
        elif record.type == "response.error":
            print(f"\n\nError: {record.error}")

    final_response = stream.get_final_response()

    print(final_response)

```
